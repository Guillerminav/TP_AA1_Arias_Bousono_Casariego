### Diagnóstico sobre el origen de las filas incompletas y decisión de tratamiento

Al inspeccionar los valores tal como están escritos en el archivo CSV, se observa una correspondencia exacta entre la presencia de datos faltantes y una anomalía en la precisión numérica:
* **Las 506 filas completas** respetan la convención histórica del dataset *Boston Housing* (entre 0 y 5 decimales según la magnitud medida).
* **Las 50 filas incompletas** presentan entre **13 y 17 decimales** en todas sus variables no binarias (incluida la variable objetivo `MEDV`), salvo la fila que está completamente vacía. Ninguna medición real de un porcentaje, distancia o precio inmobiliario se registra con esa precisión, lo que evidencia que estas 50 filas constituyen ruido sintético inyectado sobre los 506 registros originales.

**Criterio adoptado para el pipeline de preprocesamiento:**
Desde un punto de vista estrictamente estadístico, un camino válido sería descartar la totalidad de las 50 filas anómalas y trabajar únicamente con los 506 registros puros. Sin embargo, dado que uno de los objetivos pedagógicos centrales del Trabajo Práctico es evaluar e implementar técnicas de **imputación de datos faltantes** (como `KNNImputer`) cuidando rigurosamente de no cometer *Data Leakage* entre los conjuntos de entrenamiento y prueba, decidimos aplicar una **estrategia de rescate selectivo**:

1. **Eliminar la fila completamente vacía** (14 columnas nulas).
2. **Eliminar las filas donde la variable objetivo (`MEDV`) es nula** (20 filas adicionales), ya que imputar el *target* equivaldría a fabricar artificialmente el valor que el modelo debe aprender a predecir.
3. **Exigir la presencia de los predictores clave (`RM`, `TAX`, `LSTAT`, `DIS`):** Se descartan otras 17 filas que carecen de información en las variables de mayor peso predictivo (lo que barre simultáneamente con las filas más vacías que tenían entre 7 y 13 columnas nulas).
4. **Rescate e imputación controlada:** Conservamos las **12 filas incompletas remanentes** (que poseen la gran mayoría de sus datos completos) para someterlas al pipeline de escalado e imputación mediante `KNNImputer` (ajustado exclusivamente sobre `X_train`), completando así el flujo metodológico requerido por la cátedra.

```python
df_limpio = df.copy()
filas_init = len(df_limpio)

mask_vacias = df_limpio.isna().all(axis=1)
df_limpio = df_limpio[~mask_vacias]
elim_paso1 = filas_init - len(df_limpio)

print(f'Filas eliminadas por estar completamente vacías: {elim_paso1}')
```

```python
filas_antes_paso2 = len(df_limpio)
mask_sin_target = df_limpio['MEDV'].isna()
df_limpio = df_limpio[~mask_sin_target]
elim_paso2 = filas_antes_paso2 - len(df_limpio)

print(f'Filas eliminadas por no tener valor en la variable objetivo (MEDV): {elim_paso2}')
```

```python
filas_antes_paso3 = len(df_limpio)
cols_importantes = ['RM', 'TAX', 'LSTAT', 'DIS']
mask_importantes = df_limpio[cols_importantes].isna().any(axis=1)
df_limpio = df_limpio[~mask_importantes]
elim_paso3 = filas_antes_paso3 - len(df_limpio)

print(f'Filas eliminadas por no tener valor en alguna de las variables importantes {elim_paso3}')
```

# 5 - Division train/test

Generamos un split en el dataset para obtener los datos de entrenamiento (80%) y los de prueba (20%)

```python
x = df_limpio.drop(columns=['MEDV'])
y = df_limpio['MEDV']

# 1. Train-Test Split (20% para prueba)
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# 6 - Estandarización de los Datos y preprocesamiento


Aplicamos `StandardScaler` para que todas las variables numéricas tengan la misma escala (media 0 y desvío estándar 1). Ajustamos el escalador **únicamente** sobre los datos de entrenamiento (`fit`) y luego transformamos ambos conjuntos, evitando así la fuga de información estadística hacia el conjunto de prueba.

```python
scaler = StandardScaler()

# El escalador aprende SOLO del X_train y transforma ambos conjuntos
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns)
```

# Imputación de Valores Faltantes (KNN)

Utilizamos `KNNImputer` sobre los datos ya escalados para estimar las 12 filas con valores nulos remanentes basándonos en la cercanía espacial de los registros (5 vecinos). Finalmente, forzamos el redondeo en las variables categóricas u ordinales (`CHAS`, `RAD`) para que recuperen su formato original discreto.

```python
imputer = KNNImputer(n_neighbors=5)

# Imputamos aprendiendo de los vecinos del train (sobre datos escalados, KNN usa distancias)
X_train_final = pd.DataFrame(imputer.fit_transform(X_train_scaled), columns=X_train.columns)
X_test_final = pd.DataFrame(imputer.transform(X_test_scaled), columns=X_test.columns)

# Los valores imputados de CHAS y RAD pueden quedar con decimales: volvemos a la escala
# original, redondeamos y re-escalamos con el mismo scaler (ajustado solo con train)
for df_temp in [X_train_final, X_test_final]:
    original = pd.DataFrame(scaler.inverse_transform(df_temp), columns=df_temp.columns)
    original[['CHAS', 'RAD']] = original[['CHAS', 'RAD']].round()
    df_temp[:] = scaler.transform(original)

print(f"Nulos finales en Train: {X_train_final.isna().sum().sum()}")
print(f"Nulos finales en Test: {X_test_final.isna().sum().sum()}")
```

```python
# Boxplots de las variables de train ya escaladas e imputadas, para verificar el preprocesamiento
plt.figure(figsize=(16, 8))
# Excluimos CHAS (dummy) y RAD (ordinal discreta) porque el boxplot no les aporta valor
sns.boxplot(data=X_train_final.drop(columns=['CHAS', 'RAD']), orient="h", palette="Set2")
plt.title('Figura 9 - Boxplots post escalado (train)', fontweight='bold', fontsize=14)
plt.xlabel('Valor estandarizado (desvíos respecto de la media)')
plt.show()
```

Luego de aplicar StandardScaler todas las variables quedan centradas en 0 y con desvío 1.

### Decision sobre multicolinealidad TAX-RAD

```python
# VIF_j = 1 / (1 - R²_j), donde R²_j sale de explicar la variable j con las demás
def calcular_vif(X):
    vifs = {}
    for col in X.columns:
        otras = X.drop(columns=col)
        r2 = LinearRegression().fit(otras, X[col]).score(otras, X[col])
        vifs[col] = 1 / (1 - r2)
    return pd.Series(vifs).sort_values(ascending=False)

calcular_vif(X_train_final)
```

TAX y RAD tienen correlacion 0.87 y VIF alto (>5-10). Decidimos mantener ambas ya que Ridge corrige y reparte el peso entre variables correlacionadas.

# 7 - Entrenamiento de Modelos y Regularización

Entrenamos los modelos de regresión lineal múltiple en el siguiente orden: primero definimos la función de pérdida, la solución cerrada (OLS), su resolucion iterativa con gradiente descendente (batch y estocástico) y finalmente los modelos regularizados (Ridge, Lasso y ElasticNet).

```python
# Regresión Lineal Clásica
lr = LinearRegression()
lr.fit(X_train_final, y_train)

print(f'Intercepto (bias): {lr.intercept_}')
print(f'Media de y_train:  {y_train.mean()}\n')

# Coeficientes ordenados por magnitud: efecto en MEDV (k$) de aumentar 1 desvío estándar la variable
coef_ols = pd.DataFrame({'Coeficiente': lr.coef_}, index=X_train_final.columns)
coef_ols['Efecto'] = np.where(coef_ols['Coeficiente'] > 0, 'aumenta MEDV', 'disminuye MEDV')
coef_ols = coef_ols.reindex(coef_ols['Coeficiente'].abs().sort_values(ascending=False).index)
display(coef_ols.round(3))

# Adelanto del ajuste sobre train (la evaluación completa train/test está en la sección 8)
print(f'R² en train: {lr.score(X_train_final, y_train):.3f}')
```

Los coeficientes indican cuantos miles de dolares cambia MEDV al aumentar un desvio estandar esa variable con las demas fijas.
Las más influyentes son LSTAT (negativo: más poblacion de bajo nivel socioeconómico, menor precio) y RM (positivo: mas habitaciones, mayor precio).
El modelo explica ~0.7 de la variabilidad del precio en los datos de entrenamiento.

### Gradiente descendente (batch)
Minimiza el mismo MSE de forma iterativa. Para elegir el lr separamos un conjunto de validación desde el train.

```python
X_tr = X_train_final.values
y_tr = y_train.values.reshape(-1, 1)
X_te = X_test_final.values
y_te = y_test.values.reshape(-1, 1)

X_gd_tr, X_gd_val, y_gd_tr, y_gd_val = train_test_split(X_tr, y_tr, test_size=0.2, random_state=42)
```

```python
def gradiente_descendente(X_train, y_train, X_test, y_test, learning_rate=0.01, epochs=100, num_fig='10.1'):

    n = X_train.shape[0]
    m = X_train.shape[1]

    o = X_test.shape[0]

    X_train = np.hstack((np.ones((n, 1)), X_train))
    X_test = np.hstack((np.ones((o, 1)), X_test))

    W = np.random.randn(m+1).reshape(m+1, 1)

    train_errors = []
    test_errors = []

    for epoch in range(epochs):
        prediction_train = np.matmul(X_train, W)
        error_train = y_train - prediction_train
        train_mse = np.mean(error_train ** 2)
        train_errors.append(train_mse)

        # Calcular predicción y error de prueba
        prediction_test = np.matmul(X_test, W)
        error_test = y_test - prediction_test
        test_mse = np.mean(error_test ** 2)
        test_errors.append(test_mse)

        # Calcular el gradiente y actualizar pesos
        grad_sum = np.sum(error_train * X_train, axis=0)
        grad_mul = -2/n * grad_sum  # 1xm
        gradient = np.transpose(grad_mul).reshape(-1, 1)  # mx1

        W = W - (learning_rate * gradient)

  # Graficar errores de entrenamiento y prueba
    
    plt.figure(figsize=(10, 4))
    plt.plot(train_errors, label='Error de entrenamiento')
    plt.plot(test_errors, label='Error de validación')
    plt.xlabel('Época')
    plt.ylabel('Error cuadrático medio')
    plt.legend()
    plt.title(f'Figura {num_fig} - Error de entrenamiento y validación vs épocas (GD batch: lr={learning_rate}, épocas={epochs})',
              fontweight='bold', fontsize=12)
    plt.tight_layout()
    plt.show()

    return W
```

```python
np.random.seed(42) # para reproducibilidad
W_gd = gradiente_descendente(X_gd_tr, y_gd_tr, X_gd_val, y_gd_val,
                             learning_rate=0.01, epochs=1000, num_fig='10.1')

pd.DataFrame({'GD batch': W_gd.ravel(), 'LinearRegression': np.r_[lr.intercept_, lr.coef_]},
             index=['bias'] + list(X_train_final.columns))
```

### Gradiente Descendente estocástico (SGD)

```python
def stochastic_gradient_descent(X_train, y_train, X_test, y_test, lr=0.01, epochs=100):
    n = X_train.shape[0]
    m = X_train.shape[1]

    # Agregar la columna de unos para el término independiente (bias)
    X_train_b = np.hstack((np.ones((n, 1)), X_train))
    X_test_b = np.hstack((np.ones((X_test.shape[0], 1)), X_test))

    W = np.random.randn(m + 1).reshape(-1, 1)

    train_errors = []
    test_errors = []

    for i in range(epochs):
        # Permutación aleatoria para evitar sesgos
        permutation = np.random.permutation(n)
        X_train_b = X_train_b[permutation]
        y_train = y_train[permutation]

        for j in range(n):
            x_sample = X_train_b[j:j+1, :]
            y_sample = y_train[j:j+1].reshape(-1, 1)

            prediction = np.matmul(x_sample, W)
            error = y_sample - prediction
            train_mse = np.mean(error ** 2)
            train_errors.append(train_mse)

            prediction_test = np.matmul(X_test_b, W)
            error_test = y_test - prediction_test
            test_mse = np.mean(error_test ** 2)
            test_errors.append(test_mse)

            gradient = -2 * np.matmul(x_sample.T, error)
            W = W - (lr * gradient)

    plt.figure(figsize=(10, 4))
    plt.plot(train_errors, label='Error Train (1 muestra)', alpha=0.3)
    plt.plot(test_errors, label='Error de validación', alpha=0.8)
    plt.xlabel('Iteración')
    plt.ylabel('Error cuadrático medio')
    plt.legend()
    plt.title(f'Figura 11 - Convergencia SGD (Learning Rate = {lr})', fontweight='bold', fontsize=14)
    plt.show()

    return W
```

```python
np.random.seed(42) # para reproducibilidad
# Al segundo conjunto le pasamos validación, no test.
W_sgd = stochastic_gradient_descent(X_gd_tr, y_gd_tr, X_gd_val, y_gd_val, lr=0.01, epochs=10)
```

SGD baja más rapido al principio pero su curva es más irregular que la de GD batch.

# 8 - Evaluación y Comparación de Modelos

Calculamos la métrica RMSE (Raíz del Error Cuadrático Medio) sobre el conjunto de prueba aislado. Esta métrica es ideal porque penaliza los errores grandes y se expresa en la misma unidad de la variable objetivo (miles de dólares).

```python
# Función para calcular RMSE
def get_rmse(y_real, y_pred):
    return np.sqrt(mean_squared_error(y_real, y_pred))

# Predicción con los pesos W de GD/SGD (el bias está en la primera posición)
def predecir_w(X, W):
    X_b = np.hstack((np.ones((X.shape[0], 1)), np.asarray(X)))
    return (X_b @ W).ravel()

# Predicciones de cada modelo sobre train y test
predicciones = {
    'LinearRegression': (lr.predict(X_train_final), lr.predict(X_test_final)),
    'GD batch':         (predecir_w(X_train_final, W_gd), predecir_w(X_test_final, W_gd)),
    'SGD':              (predecir_w(X_train_final, W_sgd), predecir_w(X_test_final, W_sgd)),
    'Ridge':            (ridge_cv.predict(X_train_final), ridge_cv.predict(X_test_final)),
    'Lasso':            (lasso_cv.predict(X_train_final), lasso_cv.predict(X_test_final)),
    'ElasticNet':       (elastic_cv.predict(X_train_final), elastic_cv.predict(X_test_final)),
}

# RMSE en train y test para cada modelo, y la brecha entre ambos
df_comparacion = pd.DataFrame(
    {modelo: {'RMSE train': get_rmse(y_train, pred_train),
              'RMSE test': get_rmse(y_test, pred_test),
              'R2 train': r2_score(y_train, pred_train),
              'R2 test': r2_score(y_test, pred_test)}
     for modelo, (pred_train, pred_test) in predicciones.items()}
).T
df_comparacion['Brecha (test - train)'] = df_comparacion['RMSE test'] - df_comparacion['RMSE train']

# Comparación visual: RMSE de train y test lado a lado
fig, ax = plt.subplots(figsize=(10, 5))
df_comparacion[['RMSE train', 'RMSE test']].plot(kind='barh', ax=ax, color=['steelblue', 'indianred'])
ax.set_xlabel('RMSE (error en miles de dólares - menor es mejor)')
ax.set_title('Figura 12 - RMSE en entrenamiento y prueba por modelo', fontweight='bold', fontsize=14)
plt.show()

display(df_comparacion.round(3))
```

### Conclusiones Finales

1. **Ausencia de sobreajuste severo y rol de la regularización:** El `RMSE test` entre la solución analítica (`LinearRegression`: `6.311`) y los modelos regularizados (`Ridge`: `6.315`, `Lasso`: `6.305`, `ElasticNet`: `6.314`) es prácticamente idéntico. Esto demuestra que, tras la limpieza e imputación controlada, el modelo lineal clásico no sufre un problema grave de varianza, aunque `Ridge` cumple el rol clave de estabilizar los coeficientes frente a la alta colinealidad detectada entre `RAD` y `TAX` ($VIF > 5$).
2. **Regularización implícita en Gradiente Descendente (`GD batch` y `SGD`):** Ambos métodos iterativos logran un `RMSE test` ligeramente menor (`6.032` y `6.167`) y una menor brecha de generalización (`0.789` y `0.619`) que la regresión exacta (`6.311`), a pesar de tener un error levemente mayor en entrenamiento. Esto ocurre porque entrenar durante un número finito de épocas sobre una partición de validación actúa como una **detención temprana (*Early Stopping*)**, frenando la actualización de los pesos antes de que se sobreajusten al ruido específico del conjunto de entrenamiento.
3. **Límites estructurales del ajuste ($R^2 \approx 0.44 - 0.49$ en test):** El techo de desempeño de los modelos lineales en este dataset se explica por dos factores observados en el EDA: la **censura del precio en `MEDV = 50`** (que trunca artificialmente las viviendas más caras) y las **relaciones no lineales / asimetrías extremas** en variables como `LSTAT` y `CRIM`. Como futura iteración de mejora, incorporar transformaciones no lineales (como $\log(1 + \text{CRIM})$) o términos polinómicos permitiría capturar mejor la curvatura en los extremos del precio.
